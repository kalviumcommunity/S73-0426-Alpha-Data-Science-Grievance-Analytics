# Kernel Control Demonstration

This notebook demonstrates execution order, interrupting a running cell, and restarting the kernel to manage state properly.

```python
a = 10
b = 5
print("Values initialized:", a, b)
```

```python
sum_result = a + b
print("Sum is:", sum_result)
```

## Long Running Cell

The next cell simulates a long-running process that can be interrupted.

```python
import time

for i in range(1, 100):
    print("Running step:", i)
    time.sleep(1)
```

## Interrupt Demonstration

The above cell was manually interrupted before completion to stop execution safely.

## Kernel Restart Demonstration

The kernel was restarted after interrupting execution. This clears all previously stored variables.

import time

for i in range(1, 100):
    print("Running step:", i)
    time.sleep(1)

## Restart Kernel

After restarting the kernel, all variables are cleared. We need to rerun cells from the top.

```python
print(a)
```

## State Reset Verification

After restarting the kernel, variables like 'a' are no longer defined.  
This confirms that the kernel state was cleared successfully.
